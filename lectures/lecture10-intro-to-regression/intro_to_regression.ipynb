```python
# Locate the local hotel dataset and read it into a DataFrame.
from pathlib import Path


def find_data_dir():
    # Search here and then in parent folders for the project data.
    for base in (Path.cwd(), *Path.cwd().parents):
        candidate = base / "data" / "raw"
        if candidate.exists():
            # Stop at the first matching data folder.
            return candidate
    return Path("data") / "raw"


DATA_DIR = find_data_dir()

hotels = pd.read_csv(DATA_DIR / "hotels_vienna.csv")
```

### Simple Linear Regression

```python
# Add a straight-line fit as a simpler description of the price-distance association.
p1 + geom_smooth(
    aes(x="distance", y="price"), method="lm", color="blue", formula="y~x", se=False
)
```

How to quantify linear regression: \
Remember: $y = \alpha + \beta * x + \epsilon$

In Python, the [statsmodels](https://www.statsmodels.org/stable/index.html) package is usually used to estimate regressions

```python
# Load the formula-based regression interface and a percentage label formatter.
import statsmodels.formula.api as smf
from mizani.formatters import percent_format
```

We use the statsmodels formula api, where you can give the equations as a string

 Simple model, with homoskedastic SE

```python
# Fit price as a linear function of distance; fit estimates the model and summary
# reports it.
simple_reg = smf.ols("price ~ distance", data=hotels).fit()
print(simple_reg.summary())
```

Simple model, with heteroskedastic robust SE

```python
# Use HC3 robust standard errors; this changes uncertainty estimates, not the OLS
# coefficients.
hetero_rob_reg = smf.ols("price ~ distance", data=hotels).fit(cov_type ="HC3")
print(hetero_rob_reg.summary())
```

Its wasy to compare two regression output tables using the stargazer package

```python
# Load a tool for displaying several fitted models in one table.
from stargazer.stargazer import Stargazer
```

```python
# Compare the same coefficient estimates under two standard-error calculations.
table = Stargazer([simple_reg, hetero_rob_reg])
table.rename_covariates({"Intercept": "Constant"})
table.custom_columns(["Homoskedastic SE", "Heteroskedastic robust SE"], [1, 1])
table
```

## Log models 

Take log price

```python
# Take natural logs of prices; this transformation requires positive values.
hotels["lnprice"] = np.log(hotels["price"])
```

Correct distance2 measure: no closer than 0.05km

```python
# Replace very small distances with 0.05 before logging; this is an explicit modelling
# choice.
hotels["distance2"] = np.where(hotels["distance"] < 0.05, 0.05, hotels["distance"])
```

Take the log of distance2

```python
# Log the adjusted distance so zero does not produce negative infinity.
hotels["lndistance"] = np.log(hotels["distance2"])
```

Describe price and ln price

```python
# Compare summaries of price in original units and on the log scale.
hotels.filter(["price", "lnprice"]).describe(percentiles=[0.25, 0.5, 0.75, 0.95]).T
```

#### Checking and comparing all models:

```python
# Arrange the four saved plots in a two-by-two layout; the formulas define each
# specification.
g1 = pw.load_ggplot(f1 + ggtitle("Level-level"), figsize=(2, 2))
g2 = pw.load_ggplot(f2 + ggtitle("Log-level"), figsize=(2, 2))
g3 = pw.load_ggplot(f3 + ggtitle("Level-log"), figsize=(2, 2))
g4 = pw.load_ggplot(f4 + ggtitle("Log-log"), figsize=(2, 2))

glog = (g1 | g2) / (g3 | g4)
glog.savefig()
```

7. Spline

```python
# Build a continuous piecewise-linear basis; each column measures distance within a
# segment.
import copy
def lspline(series, knots):
    def knot_ceil(vector, knot):
        # Work on a copy so the original input is not changed.
        vector_copy = copy.deepcopy(vector)
        vector_copy[vector_copy > knot] = knot
        return vector_copy

    if type(knots) != list:
        knots = [knots]
    design_matrix = None
    vector = series.values

    for i in range(len(knots)):
        # print(i)
        # print(vector)
        if i == 0:
            column = knot_ceil(vector, knots[i])
        else:
            column = knot_ceil(vector, knots[i] - knots[i - 1])
        # print(column)
        if i == 0:
            design_matrix = column
        else:
            design_matrix = np.column_stack((design_matrix, column))
        # print(design_matrix)
        # Pass the unused distance on to the next segment.
        vector = vector - column
    # The final column holds distance beyond the last knot.
    design_matrix = np.column_stack((design_matrix, vector))
    # print(design_matrix)
    return design_matrix
```

```python
# Choose two miles as the point where the fitted slope may change.
cutoff = 2
```

```python
# Fit a linear spline: the slope can change at the cutoff while the fitted curve stays
# connected.
reg7 = smf.ols("price ~ lspline(distance, cutoff)", data=hotels).fit()
print(reg7.summary())
```

```python
# Draw the spline model using the same cutoff as the fitted regression.
f7 = p1 + geom_smooth(
    aes(x="distance", y="price"),
    method="lm",
    color="blue",
    formula="y ~ lspline(x, cutoff)",
    se=True,
)
f7
```

Compare these non-linear models:

```python
# Compare linear, quadratic, cubic, and spline specifications in one table.
table = Stargazer([reg1, reg5, reg6, reg7])
table.rename_covariates({"Intercept": "Constant"})
table.custom_columns(["Linear", "Squared", "Cubic", "L.Spline"], [1, 1, 1, 1])
table
```
